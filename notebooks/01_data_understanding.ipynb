```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
# set the paths
DATA_DIR = Path("../data/raw")

calendar_path = DATA_DIR / "calendar.csv"
prices_path = DATA_DIR / "sell_prices.csv"
sales_path = DATA_DIR / "sales_train_evaluation.csv"
```

```python
#Load the datasets
calendar = pd.read_csv(calendar_path)
prices = pd.read_csv(prices_path)
sales = pd.read_csv(sales_path)
```

```python
print("Calendar shape:", calendar.shape)
print("Prices shape:", prices.shape)
print("Sales shape:", sales.shape)
```

```python
display(calendar.head())
display(prices.head())
display(sales.iloc[:5, :12])
```

```python
# column names
print("Calendar columns:")
print(calendar.columns.tolist())

print("\nPrice columns:")
print(prices.columns.tolist())

print("\nSales identifier columns:")
print(sales.columns[:6].tolist())

print("\nNumber of daily sales columns:")
print(len(sales.columns[6:]))

print("\nFirst 5 daily columns:")
print(sales.columns[6:11].tolist())

print("\nLast 5 daily columns:")
print(sales.columns[-5:].tolist())
```

```python
# Check data types and memory usage of each dataset

for name, df in {
    "Calendar": calendar,
    "Prices": prices,
    "Sales": sales
}.items():
    memory_mb = df.memory_usage(deep=True).sum() / 1024**2

    print(f"\n{name}")
    print("-" * 40)
    print(df.dtypes.value_counts())
    print(f"Memory usage: {memory_mb:.2f} MB")
```

```python
print("Calendar:")
display(calendar[["date", "wm_yr_wk", "weekday", "month", "year", "d"]].dtypes)

print("\nPrices:")
display(prices.dtypes)

print("\nSales identifiers:")
display(sales.iloc[:, :6].dtypes)

print("\nSales demand columns:")
display(sales.iloc[:, 6:11].dtypes)
```

```python
print("Missing values")
print("-" * 40)

print("\nCalendar:")
print(calendar.isnull().sum())

print("\nPrices:")
print(prices.isnull().sum())

print("\nSales:")
print("Total missing values:", sales.isnull().sum().sum())
```

```python
print("Duplicate rows")
print("-" * 40)
print("Calendar:", calendar.duplicated().sum())
print("Prices:", prices.duplicated().sum())
print("Sales:", sales.duplicated().sum())

print("\nDataset coverage")
print("-" * 40)
print("Unique items:", sales["item_id"].nunique())
print("Unique departments:", sales["dept_id"].nunique())
print("Unique categories:", sales["cat_id"].nunique())
print("Unique stores:", sales["store_id"].nunique())
print("Unique states:", sales["state_id"].nunique())

print("\nPrice range")
print("-" * 40)
print(prices["sell_price"].describe())
```

```python
# Explore the product and store hierarchy in the sales data

print("Categories:")
print(sales["cat_id"].value_counts())

print("\nDepartments:")
print(sales["dept_id"].value_counts())

print("\nStores:")
print(sales["store_id"].value_counts())

print("\nStates:")
print(sales["state_id"].value_counts())
```

```python
# Measure how much historical price variation exists for each product-store combination

price_variation = (
    prices.groupby(["store_id", "item_id"])["sell_price"]
    .agg(["count", "nunique", "min", "max", "mean"])
    .reset_index()
)

price_variation["price_range"] = (
    price_variation["max"] - price_variation["min"]
)

display(price_variation.head())

print("\nProduct-store combinations:", len(price_variation))
print(
    "Combinations with multiple prices:",
    (price_variation["nunique"] > 1).sum()
)

print(
    "Percentage with multiple prices:",
    round((price_variation["nunique"] > 1).mean() * 100, 2),
    "%"
)
```

```python
# Summarize the number of unique prices and price ranges across products

display(
    price_variation[["nunique", "price_range"]]
    .describe()
)

display(
    price_variation
    .sort_values("nunique", ascending=False)
    .head(10)
)
```

```python
# Summarize daily demand values across the complete sales history

demand_data = sales.iloc[:, 6:]

print("Total units sold:", demand_data.to_numpy().sum())
print("Average daily demand:", round(demand_data.to_numpy().mean(), 2))
print("Maximum daily demand:", demand_data.to_numpy().max())

zero_percentage = (demand_data.to_numpy() == 0).mean() * 100
print("Zero-demand observations:", round(zero_percentage, 2), "%")
```

```python
# Compare overall demand levels across categories and stores

sales_summary = sales.iloc[:, :6].copy()
sales_summary["total_demand"] = demand_data.sum(axis=1)
sales_summary["avg_daily_demand"] = demand_data.mean(axis=1)

print("Demand by category:")
display(
    sales_summary.groupby("cat_id")["total_demand"]
    .sum()
    .sort_values(ascending=False)
)

print("Demand by store:")
display(
    sales_summary.groupby("store_id")["total_demand"]
    .sum()
    .sort_values(ascending=False)
)
```

```python
# Verify the common keys used to connect sales, prices, and calendar data

sales_items = set(sales["item_id"].unique())
price_items = set(prices["item_id"].unique())

sales_stores = set(sales["store_id"].unique())
price_stores = set(prices["store_id"].unique())

sales_days = set(sales.columns[6:])
calendar_days = set(calendar["d"])

print("Sales items found in prices:", sales_items.issubset(price_items))
print("Sales stores found in prices:", sales_stores.issubset(price_stores))
print("Sales days found in calendar:", sales_days.issubset(calendar_days))

print("\nUnique sales items:", len(sales_items))
print("Unique price items:", len(price_items))

print("Unique sales stores:", len(sales_stores))
print("Unique price stores:", len(price_stores))
```

## Data Understanding Summary

- The dataset contains 3,049 products across 10 stores and 3 states.
- Products belong to 7 departments and 3 main categories: FOODS, HOUSEHOLD, and HOBBIES.
- Daily sales history contains 1,941 days of demand observations.
- The price dataset contains more than 6.8 million weekly price records.
- 72.95% of product-store combinations have experienced more than one historical selling price.
- Some product-store combinations contain substantial historical price variation, making the dataset suitable for price-sensitivity analysis.
- Sales and price datasets contain no missing values or duplicate rows.
- Missing values in calendar event columns mainly represent dates without special events.
- The three datasets can be connected using product/store identifiers, calendar day identifiers, and week identifiers.
- Because converting the complete sales dataset to long format would create roughly 59 million rows, memory-efficient processing will be important in later stages.
